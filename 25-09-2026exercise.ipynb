```python
a=input()
b=a.split(",")
n=len(b)
l=[]
for i in range(n):
  for j in range(i+1,n+1):
    d=b[i:j]
    l.append(d)
e=[]
for i in l:
    f=list(set(i))
    e.append(len(f))
print(max(e))
```

```python
heights=eval(input())
water=0
for i in range(len(heights)):
    left=max(heights[:i+1])
    right=max(heights[i:])
    water+=min(left,right)-heights[i]
print(water)
```

```python
import math
b=eval(input())
n=len(b)
l=[]
f=[]
for i in range(n):
  for j in range(i+1,n+1):
    d=b[i:j]
    f.append(d)
    l.append(sum(d))
print(max(l))
for i in f:
    if(max(l)==sum(i)):
        print(i)
    
```

```python
import math
b=eval(input())
n=len(b)
l=[]
f=[]
for i in range(n):
  for j in range(i+1,n+1):
    d=b[i:j]
    f.append(d)
    l.append(math.prod(d))
print(max(l))
for i in f:
    if(max(l)==math.prod(i)):
        print(i)
    
```

```python
a=input()
b=a.split(",")
n=len(b)
m=0
ans=[]

for i in range(n):
    for j in range(i+1,n+1):
        d=b[i:j]

        if len(d)==len(set(d)):
            if len(d)>m:
                m=len(d)
                ans=d

print(ans)
print(len(ans))
```
